## Business Context
An HR analytics team wants to spot employees at higher risk of leaving (`attrition = 1`) based on tenure, overtime hours, satisfaction score, and compensation.
You will compare a single **algorithm 1** (interpretable splits, prone to overfitting when deep) against a **Algorithm 2** (bagging + random feature splits for better generalization).

### What You Will Practice
- Controlling tree depth (`max_depth`) and leaf size (`min_samples_leaf`)
- Comparing single Decision Tree vs. Random Forest on train and test metrics
- Inspecting Random Forest **feature importances**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import export_text
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

rng = np.random.default_rng(55)
n = 600

tenure_yrs = rng.uniform(0.5, 15.0, size=n).round(1)
overtime_hrs_month = rng.normal(12, 9, size=n).clip(0, 45).round(1)
job_satisfaction = rng.integers(1, 6, size=n)  # 1 (low) to 5 (high)
salary_lpa = (4.5 + 0.85 * tenure_yrs + rng.normal(0, 1.8, size=n)).clip(3.0, 28.0).round(2)

logit = (
    -0.22 * tenure_yrs
    + 0.065 * (overtime_hrs_month - 12)
    - 0.45 * (job_satisfaction - 3)
    - 0.08 * (salary_lpa - 8.0)
    - 0.8
)
prob_leave = 1 / (1 + np.exp(-logit))
attrition = (rng.random(n) < prob_leave).astype(int)

df_hr = pd.DataFrame({
    'tenure_yrs': tenure_yrs,
    'overtime_hrs_month': overtime_hrs_month,
    'job_satisfaction': job_satisfaction,
    'salary_lpa': salary_lpa,
    'attrition': attrition
})

print('Attrition rate:', f"{df_hr['attrition'].mean():.1%}")
print(df_hr.head().to_string(index=False))
```
