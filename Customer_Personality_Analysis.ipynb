**Reasoning**:
The first step is to load the data into a pandas DataFrame and display its initial rows and information.



```python
import pandas as pd

df = pd.read_csv('/content/marketing_campaign.csv', sep='\t')
display(df.head())
display(df.info())
```

## Explore the data

### Subtask:
Perform initial data exploration to understand the structure, content, and quality of the data. This includes checking data types, missing values, and summary statistics.


**Reasoning**:
Display the number of unique values in each column, calculate and display descriptive statistics for numerical columns, and identify and display the count of missing values for each column.



```python
display(df.nunique())
display(df.describe())
display(df.isnull().sum())
```

## Preprocess the data

### Subtask:
Clean and preprocess the data for analysis. This might involve handling missing values, transforming variables, and creating new features.


**Reasoning**:
Handle missing values, create new features, and drop unnecessary columns as per the instructions.



```python
# 1. Handle missing values in 'Income'
df['Income'].fillna(df['Income'].mean(), inplace=True)

# 2. Create 'Age' feature
current_year = 2025
df['Age'] = current_year - df['Year_Birth']

# 3. Create 'Total_Spending' feature
spending_cols = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
df['Total_Spending'] = df[spending_cols].sum(axis=1)

# 4. Drop original columns
df.drop(['Year_Birth', 'Z_CostContact', 'Z_Revenue'], axis=1, inplace=True)

# 5. One-hot encode 'Education' and 'Marital_Status'
df = pd.get_dummies(df, columns=['Education', 'Marital_Status'], drop_first=True)

display(df.head())
display(df.info())
```

## Perform customer segmentation

### Subtask:
Apply clustering techniques or other methods to segment customers based on their characteristics and behaviors.


**Reasoning**:
Select relevant numerical features, standardize them, apply K-Means clustering, and add cluster labels to the DataFrame.



```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Select relevant numerical features for clustering
numerical_features = ['Income', 'Age', 'Total_Spending', 'MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
X = df[numerical_features]

# Standardize the selected features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10) # Choosing 4 clusters as a starting point
df['Cluster'] = kmeans.fit_predict(X_scaled)

# Display the first few rows with the new 'Cluster' column
display(df.head())
```

## Analyze segments

### Subtask:
Analyze the characteristics of each customer segment to understand their unique profiles and preferences.


**Reasoning**:
Group the DataFrame by 'Cluster' and calculate the mean of numerical features and sum of one-hot encoded features for each cluster, storing the results in separate dataframes. Then display both dataframes.



```python
cluster_analysis = df.groupby('Cluster')[numerical_features].mean()
display(cluster_analysis)

categorical_features = [col for col in df.columns if col.startswith(('Education_', 'Marital_Status_'))]
categorical_analysis = df.groupby('Cluster')[categorical_features].sum()
display(categorical_analysis)
```

**Reasoning**:
Based on the cluster analysis, write a brief description for each cluster to highlight their key characteristics and differences.



```python
print("Cluster 0:")
print("- Moderate income (~$66k) and older age (~59).")
print("- High total spending, particularly on wines and meat products.")
print("- Predominantly have Graduation or PhD education.")
print("- Mix of marital statuses, with a notable number of married and together individuals.")
print("\nCluster 1:")
print("- Highest income (~$77k) and middle age (~56).")
print("- Very high total spending, especially on wines, meat, and fish products.")
print("- Majority have Graduation education.")
print("- Relatively balanced across marital statuses, with slightly more married individuals.")
print("\nCluster 2:")
print("- Lowest income (~$34k) and youngest age (~48).")
print("- Very low total spending across all categories.")
print("- Diverse educational backgrounds, with a significant number in Graduation and Master's.")
print("- Highest number of single and married individuals.")
print("\nCluster 3:")
print("- Moderate-low income (~$46k) and oldest age (~67).")
print("- Low total spending, slightly more on wines compared to other products.")
print("- Mix of educational backgrounds, with more in Graduation and PhD.")
print("- Higher proportion of married and together individuals.")
```

## Visualize results

### Subtask:
Create visualizations to represent the customer segments and their key attributes.


**Reasoning**:
Generate the requested visualizations to represent the customer segments and their key attributes. This involves creating a scatter plot, two bar plots, and ensuring they have appropriate labels and titles.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Scatter plot of 'Income' vs. 'Total_Spending', colored by 'Cluster'
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Income', y='Total_Spending', hue='Cluster', palette='viridis', s=50)
plt.title('Customer Segmentation: Income vs. Total Spending')
plt.xlabel('Income')
plt.ylabel('Total Spending')
plt.grid(True)
plt.show()

# 2. Bar plot showing the average 'Total_Spending' for each cluster
plt.figure(figsize=(8, 5))
sns.barplot(x=cluster_analysis.index, y='Total_Spending', data=cluster_analysis, palette='viridis')
plt.title('Average Total Spending per Cluster')
plt.xlabel('Cluster')
plt.ylabel('Average Total Spending')
plt.show()

# 3. Bar plot showing the distribution of 'Education' levels within each cluster
# Reshape the categorical_analysis DataFrame for easier plotting
education_cols = [col for col in categorical_analysis.columns if col.startswith('Education_')]
education_analysis = categorical_analysis[education_cols].stack().reset_index(name='Count')
education_analysis.rename(columns={'level_1': 'Education'}, inplace=True)

plt.figure(figsize=(12, 6))
sns.barplot(data=education_analysis, x='Cluster', y='Count', hue='Education', palette='viridis')
plt.title('Education Distribution within Clusters')
plt.xlabel('Cluster')
plt.ylabel('Number of Customers')
plt.show()
```
