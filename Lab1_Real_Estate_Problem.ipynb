## Business Context
A real estate advisory platform across metro cities (like Bengaluru, Pune, and Hyderabad) wants to estimate apartment prices in **₹ Lakhs** based on property size, configuration, age, and metro proximity.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

rng = np.random.default_rng(21)
n = 500

area_sqft = rng.normal(1250, 380, size=n).clip(500, 2800)
bedrooms = np.where(area_sqft < 850, 1, np.where(area_sqft < 1450, 2, 3))
building_age_yrs = rng.integers(0, 20, size=n)
metro_dist_km = rng.exponential(scale=4.0, size=n).clip(0.3, 20.0)

# Price in INR Lakhs
price_lakhs = (
    18.0
    + 0.055 * area_sqft
    + 8.5 * (bedrooms - 2)
    - 0.75 * building_age_yrs
    - 1.6 * metro_dist_km
    + rng.normal(0, 12.0, size=n)
).clip(25.0, 260.0)

df_house = pd.DataFrame({
    'area_sqft': np.round(area_sqft, 0).astype(int),
    'bedrooms': bedrooms,
    'building_age_yrs': building_age_yrs,
    'metro_dist_km': np.round(metro_dist_km, 1),
    'price_lakhs': np.round(price_lakhs, 2)
})

print('Shape:', df_house.shape)
print('Missing values:', df_house.isna().sum().to_dict())
print(df_house.head().to_string(index=False))
```

## Step 2: Train-Test Split, Linear Regression & Evaluation
- **MAE (Mean Absolute Error):** Average absolute error in ₹ Lakhs.
- **RMSE (Root Mean Squared Error):** Penalizes larger errors more heavily (same units as target: ₹ Lakhs).
- **$R^2$ (Coefficient of Determination):** Share of variance in `price_lakhs` explained by the model.
