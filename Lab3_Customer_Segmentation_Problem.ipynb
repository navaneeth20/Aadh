Lab 3: Retail Customer Segmentation

## Business Context
An e-commerce / retail brand has no pre-labeled customer segments and wants to group shoppers by **annual spend (₹)**, **purchase frequency**, and **average basket size** to tailor loyalty campaigns.
Unlike Sessions 2–3, there is **no target label `y`**—this is **? learning**.

### What You Will Practice
- Scaling features before distance-based clustering
- Evaluating cluster counts using **Inertia (Elbow)** and **Silhouette Score**
- Profiling clusters in original units so marketing/business teams can act on them

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import silhouette_score

rng = np.random.default_rng(44)

# Three natural shopper personas
g1 = pd.DataFrame({
    'annual_spend_inr': rng.normal(12000, 3500, 150).clip(3000, 25000),
    'orders_per_year': rng.normal(4, 1.5, 150).clip(1, 10),
    'avg_order_inr': rng.normal(3000, 700, 150).clip(1000, 6000)
})
g2 = pd.DataFrame({
    'annual_spend_inr': rng.normal(45000, 9000, 150).clip(20000, 75000),
    'orders_per_year': rng.normal(15, 3.5, 150).clip(8, 28),
    'avg_order_inr': rng.normal(3100, 650, 150).clip(1500, 5500)
})
g3 = pd.DataFrame({
    'annual_spend_inr': rng.normal(95000, 18000, 100).clip(60000, 160000),
    'orders_per_year': rng.normal(11, 3.0, 100).clip(5, 22),
    'avg_order_inr': rng.normal(8800, 1800, 100).clip(5000, 15000)
})

df_seg = pd.concat([g1, g2, g3], ignore_index=True).sample(frac=1.0, random_state=42).reset_index(drop=True)
df_seg = df_seg.round({'annual_spend_inr': 0, 'orders_per_year': 1, 'avg_order_inr': 0})
print('Total customers:', len(df_seg))
print(df_seg.head().to_string(index=False))
```
