Lab 1: Retail Loan Approval (Linear Regression)

## Business Context
A lending team evaluates personal and consumer loan applications using applicant income, credit score, debt-to-income ratio, and employment tenure.
**Algorithm** predicts a probability between `0` and `1` using the sigmoid curve and applies a decision threshold (default `0.5`) to classify `Approved (1)` vs. `Rejected (0)`.

### What You Will Practice
- Training `LinearRegression` on scaled tabular features
- Inspecting `predict_proba` vs. `predict`
- Reading **Confusion Matrix, Precision, Recall, and F1-score**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

rng = np.random.default_rng(33)
n = 550

monthly_income_k = rng.normal(75, 32, size=n).clip(20, 250)  # in ₹ thousands
credit_score = rng.normal(720, 60, size=n).clip(500, 850)
debt_to_income = rng.uniform(0.08, 0.75, size=n)
job_tenure_yrs = rng.exponential(scale=3.5, size=n).clip(0.2, 20.0)

score = (
    0.025 * (monthly_income_k - 75)
    + 0.022 * (credit_score - 720)
    - 4.2 * (debt_to_income - 0.35)
    + 0.14 * job_tenure_yrs
    + 0.2
)
prob_approve = 1 / (1 + np.exp(-score))
approved = (rng.random(n) < prob_approve).astype(int)

df_loan_clf = pd.DataFrame({
    'monthly_income_k': np.round(monthly_income_k, 1),
    'credit_score': np.round(credit_score, 0).astype(int),
    'debt_to_income': np.round(debt_to_income, 2),
    'job_tenure_yrs': np.round(job_tenure_yrs, 1),
    'approved': approved
})

print('Approval rate:', f"{df_loan_clf['approved'].mean():.1%}")
print(df_loan_clf.head().to_string(index=False))
```
